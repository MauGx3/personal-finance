## 3. Technical Analysis Indicators

Let's calculate various technical indicators using sample price data.

```python
# Generate sample price data
np.random.seed(42)  # For reproducible results
dates = pd.date_range('2024-01-01', periods=100, freq='D')

# Simulate realistic stock price movement
base_price = 100
returns = np.random.normal(0.001, 0.02, 100)  # Daily returns with slight upward bias
prices = [base_price]
for ret in returns[1:]:
    prices.append(prices[-1] * (1 + ret))

close_prices = pd.Series(prices, index=dates)
high_prices = close_prices + np.random.uniform(0.5, 2.0, 100)
low_prices = close_prices - np.random.uniform(0.5, 2.0, 100)
volume = pd.Series(np.random.randint(1_000_000, 10_000_000, 100), index=dates)

print(f"Sample price data generated: {len(close_prices)} days")
print(f"Price range: ${low_prices.min():.2f} - ${high_prices.max():.2f}")
print(f"Final price: ${close_prices.iloc[-1]:.2f}")
```

```python
# Calculate moving averages
sma_20 = TechnicalIndicators.moving_average(close_prices, 20)
sma_50 = TechnicalIndicators.moving_average(close_prices, 50)
ema_12 = TechnicalIndicators.exponential_moving_average(close_prices, 12)

print("📈 MOVING AVERAGES")
print(f"20-day SMA: ${sma_20.iloc[-1]:.2f}")
print(f"50-day SMA: ${sma_50.iloc[-1]:.2f}")
print(f"12-day EMA: ${ema_12.iloc[-1]:.2f}")
print(f"Price vs 20-day SMA: {((close_prices.iloc[-1] / sma_20.iloc[-1]) - 1):.1%}")
```

```python
# Calculate momentum indicators
rsi = TechnicalIndicators.rsi(close_prices)
macd_data = TechnicalIndicators.macd(close_prices)
stoch_data = TechnicalIndicators.stochastic_oscillator(high_prices, low_prices, close_prices)
williams_r = TechnicalIndicators.williams_r(high_prices, low_prices, close_prices)

print("⚡ MOMENTUM INDICATORS")
print(f"RSI (14-day): {rsi.iloc[-1]:.1f}")
print(f"MACD: {macd_data['macd'].iloc[-1]:.3f}")
print(f"MACD Signal: {macd_data['signal'].iloc[-1]:.3f}")
print(f"Stochastic %K: {stoch_data['%K'].iloc[-1]:.1f}")
print(f"Stochastic %D: {stoch_data['%D'].iloc[-1]:.1f}")
print(f"Williams %R: {williams_r.iloc[-1]:.1f}")

# Interpret RSI
rsi_value = rsi.iloc[-1]
if rsi_value > 70:
    rsi_signal = "Overbought ⚠️"
elif rsi_value < 30:
    rsi_signal = "Oversold 🚀"
else:
    rsi_signal = "Neutral ➡️"
    
print(f"RSI Signal: {rsi_signal}")
```

```python
# Calculate volatility indicators
bb_data = TechnicalIndicators.bollinger_bands(close_prices)
atr = TechnicalIndicators.average_true_range(high_prices, low_prices, close_prices)

print("📊 VOLATILITY INDICATORS")
print(f"Bollinger Upper: ${bb_data['upper'].iloc[-1]:.2f}")
print(f"Bollinger Middle: ${bb_data['middle'].iloc[-1]:.2f}")
print(f"Bollinger Lower: ${bb_data['lower'].iloc[-1]:.2f}")
print(f"Average True Range: ${atr.iloc[-1]:.2f}")

# Bollinger Band position
current_price = close_prices.iloc[-1]
bb_upper = bb_data['upper'].iloc[-1]
bb_lower = bb_data['lower'].iloc[-1]
bb_position = (current_price - bb_lower) / (bb_upper - bb_lower)

print(f"Bollinger Band Position: {bb_position:.1%}")
if bb_position > 0.8:
    print("Position: Near upper band (potential resistance) 🔴")
elif bb_position < 0.2:
    print("Position: Near lower band (potential support) 🟢")
else:
    print("Position: Within normal range 🟡")
```

```python
# Calculate volume indicators
vwap = TechnicalIndicators.volume_weighted_average_price(high_prices, low_prices, close_prices, volume)
obv = TechnicalIndicators.on_balance_volume(close_prices, volume)
mfi = TechnicalIndicators.money_flow_index(high_prices, low_prices, close_prices, volume)

print("📊 VOLUME INDICATORS")
print(f"VWAP: ${vwap.iloc[-1]:.2f}")
print(f"OBV: {obv.iloc[-1]:,.0f}")
print(f"Money Flow Index: {mfi.iloc[-1]:.1f}")
print(f"Price vs VWAP: {((current_price / vwap.iloc[-1]) - 1):.1%}")
```

## 4. Advanced Technical Analysis: Ichimoku Cloud

The Ichimoku Cloud provides a complete trading system.

```python
# Calculate Ichimoku Cloud
ichimoku = TechnicalIndicators.ichimoku_cloud(high_prices, low_prices, close_prices)

print("☁️ ICHIMOKU CLOUD ANALYSIS")
print(f"Tenkan-sen (Conversion Line): ${ichimoku['tenkan_sen'].iloc[-1]:.2f}")
print(f"Kijun-sen (Base Line): ${ichimoku['kijun_sen'].iloc[-1]:.2f}")
print(f"Senkou Span A: ${ichimoku['senkou_span_a'].iloc[-1]:.2f}")
print(f"Senkou Span B: ${ichimoku['senkou_span_b'].iloc[-1]:.2f}")

# Analyze Ichimoku signals
tenkan = ichimoku['tenkan_sen'].iloc[-1]
kijun = ichimoku['kijun_sen'].iloc[-1]
senkou_a = ichimoku['senkou_span_a'].iloc[-1]
senkou_b = ichimoku['senkou_span_b'].iloc[-1]

print("\n🎯 ICHIMOKU SIGNALS:")
if tenkan > kijun:
    print("✅ Tenkan-Kijun Cross: Bullish (Tenkan above Kijun)")
else:
    print("❌ Tenkan-Kijun Cross: Bearish (Tenkan below Kijun)")

cloud_top = max(senkou_a, senkou_b)
cloud_bottom = min(senkou_a, senkou_b)

if current_price > cloud_top:
    print("🚀 Price position: Above cloud (bullish)")
elif current_price < cloud_bottom:
    print("🔻 Price position: Below cloud (bearish)")
else:
    print("☁️ Price position: Inside cloud (neutral/consolidation)")
```
